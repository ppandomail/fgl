## Conceptos básicos

1. ¿Qué es un compilador?
1. ¿Qué diferencia hay entre un compilador y un intérprete?
1. ¿Cuales son las etapas principales?
1. ¿Por qué separar lexer y parser?
    * porque cumplen funciones distintas: lexer reconoce palabras (tokens) y parser verifica la estructura. Esta separación hace el compilador más modular y fácil de mantener
1. ¿Puede existir un parser sin lexer?
    * sí, pero sería poco práctico. El parser tendría que analizar carácter por carácter, haciendo el proceso más complejo

## Scanner

1. ¿Qué diferencia hay entre carácter y token?
    * caracter es un símbolo y token es la unidad reconocida
1. ¿Por qué el léxico usa expresiones regulares?
    * porque los tokens forman un lenguaje regular
1. ¿Qué hace el lexer?
    * convierte caracteres en tokens usando expresiones regulares
1. ¿Qué ocurre si aparece un carácter desconocido?
    * el lexer genera un error léxico (def t_error(t))

## Parser

1. ¿Qué recibe el parser?
    * tokens
1. ¿Qué hace el parser?
    verifica que la secuencia de tokens cumpla la gramática y construye la estructura sintáctica
1. ¿Qué ocurre si una gramática es ambigua?
    * una misma cadena puede tener más de un árbol sintáctico. Esto suele producir conflictos en el parser y debe resolverse modificando la gramática o definiendo precedencia y asociatividad
1. ¿Qué devuelve el parser?
    * un árbol sintáctico / un resultado / una estructura de datos
1. ¿Qué es Backtracking?
    * es la técnica en la que un parser prueba una producción y, si no conduce a una derivación válida, retrocede para intentar otra alternativa
1. ¿Todos los parsers descendentes usan Backtracking?
    * No. Los predictivos (LL(1)) no lo necesitan

## Errores

1. ¿Qué diferencia hay entre error léxico y sintáctico?
    * Léxico: existe un carácter inválido
    * Sintáctico: los tokens están mal ordenados
1. ¿Puede un lexer detectar errores de sintaxis?
    * no. Solo detecta errores relacionados con caracteres o tokens inválidos. La estructura de la entrada la valida el parser
